# Tokenization — Vocabulary pruning after extension (shrink the embedding table) (`core/tokenization`)

```python
from aligntune.core.backend_factory import create_tokenization_trainer

# Pruning is a modifier on top of an extension method: after adding new
# tokens, leaf-based pruning removes redundant/unreachable ones to shrink the
# embedding table before deployment. Leaf-based is the structure-safe method
# -- it doesn't break the BPE merge graph the way naive frequency/last-N do.
trainer = create_tokenization_trainer(
    base_model="gpt2",
    target_languages=["zh"],
    dataset_name="wikimedia/wikipedia",
    config_name="20231101.zh",
    num_new_tokens=200,
    extension_method="continued_bpe",
    output_dir="./out_tokenizer_zh_pruned",
    max_samples=200,
    prune=True,
    pruning_ratio=0.1,
    pruning_method="leaf_frequency",
)

result = trainer.train()
print("Vocabulary extension + pruning completed.")
print(result)
```
