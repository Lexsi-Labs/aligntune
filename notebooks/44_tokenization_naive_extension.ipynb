# Tokenization — Naive vocabulary extension (fast prototyping) (`core/tokenization`)

```python
from aligntune.core.backend_factory import create_tokenization_trainer

# Naive extension trains a fresh tokenizer on the new corpus and diffs it
# against the base vocabulary -- faster to try, but a lower quality bar than
# continued BPE (more unreachable/wasted tokens).
trainer = create_tokenization_trainer(
    base_model="gpt2",
    target_languages=["zh"],
    dataset_name="wikimedia/wikipedia",
    config_name="20231101.zh",
    num_new_tokens=10000,
    extension_method="naive_extension",
    output_dir="./out_tokenizer_zh_naive",
    max_samples=1000,
)

result = trainer.train()
print("Naive vocabulary extension completed.")
print(result)
```
