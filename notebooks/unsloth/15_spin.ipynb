# SPIN (Self-Play Fine-Tuning) — Qwen2.5-0.5B-Instruct on Alpaca (TRL backend)

```python
from aligntune.core.backend_factory import create_rl_trainer

# NOTE: SPIN generates one full opponent response per training example
# up-front (before any DPO round starts), sequentially - unlike the GRPO-
# family trainers which only generate per-batch during each step. At
# max_samples=1000 this pre-generation phase alone can take well over an
# hour on a single T4. Verified correct and crash-free end-to-end on the
# unsloth backend with max_samples=8 (~2s total); reduce max_samples below
# if you want a fast run.
trainer = create_rl_trainer(
    model_name="Qwen/Qwen2.5-0.5B-Instruct",
    dataset_name="tatsu-lab/alpaca",
    split="train",
    algorithm="spin",
    backend="unsloth",
    output_dir="./out_spin",
    num_epochs=1,
    max_steps=2,
    batch_size=2,
    learning_rate=1e-4,
    max_seq_length=256,
    max_samples=8,
    max_prompt_length=128,
    use_peft=True,
    lora_r=8,
    num_rounds=1,
    generation_max_length=128,
    eval_strategy="no",
    save_strategy="no",
    report_to="none",
    seed=42,
)

results = trainer.train()
print("SPIN training completed.")
print(results)
```
