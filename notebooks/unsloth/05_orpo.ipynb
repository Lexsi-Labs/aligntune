# ORPO — Qwen2.5-0.5B on Anthropic/hh-rlhf (TRL backend)

```python
from aligntune.core.backend_factory import create_rl_trainer

trainer = create_rl_trainer(
    model_name="Qwen/Qwen2.5-0.5B",
    dataset_name="Anthropic/hh-rlhf",
    algorithm="orpo",
    backend="unsloth",
    output_dir="./out_orpo",
    max_steps=10,
    batch_size=2,
    learning_rate=5e-6,
    max_seq_length=384,
    split='train',
    max_samples=1000,
)

results = trainer.train()
print("ORPO training completed.")
print(results)
```
