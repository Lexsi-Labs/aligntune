# Online DPO — Qwen2.5-0.5B-Instruct on Anthropic/hh-rlhf (TRL backend, self-reward)

```python
from aligntune.core.backend_factory import create_rl_trainer

# Online DPO requires an explicit reward source -- it no longer silently
# defaults to reusing the policy model as its own (untrained) reward model.
trainer = create_rl_trainer(
    model_name="Qwen/Qwen2.5-0.5B-Instruct",
    dataset_name="Anthropic/hh-rlhf",
    algorithm="online_dpo",
    split = 'train',
    backend="unsloth",
    output_dir="./out_dpo",
    max_steps=3,
    batch_size=4,
    learning_rate=5e-6,
    max_seq_length=768,
    max_samples=16,
    reward_value_loading_type="unsloth",
    reward_model_name="Qwen/Qwen2.5-0.5B-Instruct",
)

results = trainer.train()
print("DPO training completed.")
print(results)
```
