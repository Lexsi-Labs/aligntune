# DPO — Qwen2.5-0.5B-Instruct on Anthropic/hh-rlhf (TRL backend)

```python
from aligntune.core.backend_factory import create_rl_trainer

trainer = create_rl_trainer(
    model_name="Qwen/Qwen2.5-0.5B-Instruct",
    dataset_name="Anthropic/hh-rlhf",
    algorithm="dpo",
    split = 'train',
    backend="unsloth",
    output_dir="./out_dpo",
    max_steps=10,
    batch_size=4,
    learning_rate=5e-6,
    max_seq_length=256,
    max_samples=1000,
)

results = trainer.train()
print("DPO training completed.")
print(results)
```
